### Artificial Neural Networks (ANN)

```python
# Artificial Neural Network (ANN)
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical


#load dataset

(x_train, y_train), (x_test, y_test) = mnist.load_data()


#normalise data (0-255 / 0-1)
x_train = x_train / 255.0
x_test = x_test / 255.0

y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

#Build the ANN model 
model=Sequential([
    Flatten(input_shape=(28,28)),                  #Flatten 28*28 image to 784 input nuerons
    Dense(128, activation='relu'),                #Hidden layer with 128 nuerons and relu
    Dense(10, activation='softmax')               #output layer with 10 nuerons(for 10 classes)

])             



#Complie the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

#train the model
model.fit(x_train, y_train, epochs=50, batch_size=32, validation_split=0.1)



#Evaluate the test set
loss, accuracy = model.evaluate(x_test, y_test)
print("Test accuracy:", accuracy)
```
